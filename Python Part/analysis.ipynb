```python
RAW_PATH = "Bengaluru_Ola.csv"
OUT_PATH = "Bengaluru_Ola_cleaned.csv"
```

```python
df['Booking Status'].value_counts
```

```python
df.columns = [c.strip().replace("  ", " ") for c in df.columns]
```

```python
df.columns
```

```python
df['ride_datetime'] =  pd.to_datetime(df['Date'] + " " + df['Time'], format = "%d/%m/%Y %H:%M:%S")
```

```python
df['ride_datetime'].head()
```

```python
df['day_of_week'] = df['ride_datetime'].dt.day_name()
```

```python
df['hour'] = df['ride_datetime'].dt.hour
```

```python
df['hour']
```

```python
df['Booking ID'].duplicated().sum()
```

```python
df[df['Booking ID'].duplicated(keep=False)].sort_values('Booking ID')
```

```python
df['ride_id'] = 'RIDE' + df.index.astype(str).str.zfill(6)
```

```python
df['ride_id']
```

```python
df['is_completed'] = df['Booking Status'] == 'Success'
```

```python
reason_cols = ['Reason for Cancelling by Customer', 'Reason for Cancelling by Driver', 'Incomplete Rides Reason']
for col in reason_cols:
    df[col] = df[col].fillna('Not Applicable')
```

```python
def area_to_zone(area_str):
    area_num = int(area_str.split("-")[1])
    zone_num = ((area_num - 1) * 8 // 50) + 1
    return f"Zone-{zone_num}"
```

```python
df["pickup_zone"] = df["Pickup Location"].apply(area_to_zone)
df["drop_zone"] = df["Drop Location"].apply(area_to_zone)
```

```python
df.to_csv(OUT_PATH, index=False)
```

```python
naive_avg = pd.read_csv('Bengaluru Ola.csv')["Booking Value"].fillna(0).mean()
correct_avg = df.loc[df["is_completed"], "Booking Value"].mean()
error_pct = abs(naive_avg - correct_avg) / correct_avg * 100
 
print("Rows before:", len(pd.read_csv('Bengaluru Ola.csv')))
print("Rows after:", len(df))
print("Duplicate Booking IDs fixed:", pd.read_csv('Bengaluru Ola.csv')["Booking ID"].duplicated().sum())
print("Zones created:", df["pickup_zone"].nunique())
print(f"Accuracy error avoided by not zero-filling: {error_pct:.1f}%")
```
